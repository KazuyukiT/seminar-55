## 計算と変数

### 計算を表示する

````python
print(1200 + 800)
print(1200 * 3)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 3)
````

**結果（上から）：** `2000`、`3600`、`3.5`、`3`、`1`、`8`。

`*` は掛け算、`/` は割り算、`//` は商を小さい整数側へ丸める除算、`%` は余り、`**` は累乗です。掛け算・割り算は足し算より先に計算されます。順序を変えたいときは括弧を使います。

### 値に名前を付ける

````python
price = 1200
quantity = 3
total = price * quantity
print(total)
````

**結果：** `3600`。`price` などを変数と呼びます。`=` は「右側を計算し、左側の名前に結び付ける」という代入です。数学の等式とは意味が異なります。

````python
quantity = 4
print(total)
total = price * quantity
print(total)
````

**結果：** `3600`、`4800`。変数を変えても、過去に計算した値は自動更新されません。計算し直す必要があります。

変数名には `study_hours` のように意味の分かる英小文字とアンダースコアを使います。大文字・小文字は区別されます。`print`、`sum`、`list` のような組み込み機能の名前は変数名に使わないでください。

| 型 | 例 | 用途 |

`#` から行末はコメントです。処理の目的や単位をメモできます。

**結果：** `True`。`>=` は「以上」の比較です。

## リスト：複数の値をまとめる

5人の1週間の学習時間を、リストにまとめます。角括弧 `[]` で囲み、値をコンマで区切ります。この章のデータはすべて架空です。

````python
study_hours = [2, 5, 0, 3, 10]
print(study_hours[0])
print(study_hours[-1])
print(study_hours[1:3])
print(len(study_hours))
````

**結果：** `2`、`10`、`[5, 0]`、`5`。位置を表す番号（インデックス）は0から始まります。`-1` は最後、`[1:3]` は1番目から3番目の直前までです。`len()` は要素数を返します。

````python
total_hours = sum(study_hours)
mean_hours = total_hours / len(study_hours)
print(total_hours)
print(mean_hours)
print(min(study_hours), max(study_hours))
````

**結果：** `20`、`4.0`、`0 10`。`sum()` は合計、`min()` と `max()` は最小値・最大値です。平均は空のリストでは計算できません。

## 条件分岐：条件で処理を変える

````python
hours = 5
if hours >= 5:
    print("5時間以上")
else:
    print("5時間未満")
````

**結果：** `5時間以上`。`if` の後ろに条件を書き、行末に `:` を付けます。条件が成り立つと、その下の字下げした処理が実行されます。`else` はそれ以外です。字下げは半角スペース4個にそろえます。

| 比較 | 意味 |
|---|---|
| `a == b` | 等しい |
| `a != b` | 等しくない |
| `a > b`、`a >= b` | より大きい、以上 |
| `a < b`、`a <= b` | より小さい、以下 |

代入の `=` と比較の `==` を区別してください。条件を組み合わせるときは `and`（両方）、`or`（少なくとも一方）を使います。

````python
hours = 0
if hours == 0:
    print("0時間")
elif hours < 5:
    print("0時間より多く5時間未満")
else:
    print("5時間以上")
````

**結果：** `0時間`。`elif` は追加の条件です。上から調べ、最初に成り立った枝だけを実行します。この例は0以上の時間を前提にしています。

### 演習4：境界を確かめる

最初の `if` の例を、`hours` が4、5、6の場合で試してください。条件を `hours > 5` に変えると、どの場合だけ結果が変わりますか。

## 繰り返し：同じ処理を各データに行う

````python
study_hours = [2, 5, 0, 3, 10]
for hours in study_hours:
    print(hours * 60)
````

**結果（上から）：** `120`、`300`、`0`、`180`、`600`。`for` はリストから値を順に取り出して `hours` に代入し、字下げした処理を繰り返します。

### 条件に合う人数を数える

````python
count = 0
for hours in study_hours:
    if hours >= 5:
        count = count + 1
print(count)
````

**結果：** `2`。`count = count + 1` は、現在の値に1を足して代入し直します。`count` の初期化はループの前、最後の表示はループの外です。実行前に、2、5、0、3、10を読んだ直後の `count` を紙に書いてみてください。

`range(3)` は0、1、2を順に生成します。決まった回数の繰り返しで使えます。

````python
for i in range(3):
    print(i)
````

## 辞書：項目名と値を対応させる

辞書は、項目名（キー）で値を取り出す仕組みです。リストの「何番目」と区別しましょう。

````python
student = {"id": "A", "hours": 2}
print(student["id"])
print(student["hours"])
````

**結果：** `A`、`2`。辞書をリストに入れると、小さな表のように複数人の情報を扱えます。

````python
students = [
    {"id": "A", "hours": 2},
    {"id": "B", "hours": 5},
    {"id": "C", "hours": 0},
]
for student in students:
    print(f"{student['id']}: {student['hours']}時間")
````

**結果：** `A: 2時間`、`B: 5時間`、`C: 0時間`。外側の二重引用符と内側の一重引用符を区別してください。大量の表データを扱うpandasは、この基礎の後で学びます。

## 関数：処理に名前を付ける

同じ計算を別の値でも使えるよう、関数にまとめます。

````python
def to_minutes(hours):
    return hours * 60

result = to_minutes(2.5)
print(result)
````

**結果：** `150.0`。`def` は関数の定義、`hours` は受け取る値の名前（引数）、`return` は返す値（戻り値）の指定です。定義しただけでは計算は実行されません。`to_minutes(2.5)` と呼び出すと動きます。

`print()` は表示、`return` は呼び出し元へ値を渡す操作です。返された値は次の計算に使えます。

````python
print(to_minutes(1) + to_minutes(2))
````

**結果：** `180`。関数内で `return` の代わりに `print` だけを使うと、呼び出しの戻り値は `None`（値がないことを表す値）になります。

## 総合課題：学習時間の集計メモ

次のデータを新しいセルに定義してください。これは小規模な架空データであり、実際の学生集団について推論するための標本ではありません。

1. 人数、合計、平均を計算し、単位とともに表示します。
2. 5時間以上の人数と割合（%）を求めます。人数は `for` と `if` で数え、割合は演習6の関数を使います。

### 総合課題の解答例

````python
from statistics import median

def percentage(part, whole):
    return part / whole * 100

survey_hours = [2, 5, 0, 3, 10]
n = len(survey_hours)
total = sum(survey_hours)
count = 0
for hours in survey_hours:
    if hours >= 5:
        count = count + 1

print(f"人数: {n}人")
print(f"合計: {total}時間")
print(f"平均: {total / n:.1f}時間")
print(f"5時間以上: {count}人")
print(f"割合: {percentage(count, n):.1f}%")
print(f"中央値: {median(survey_hours)}時間")
````

`:.1f` は小数点以下を1桁にそろえて表示する指定です。このコードは空でない数値リストを前提にしています。

考察例：平均は4.0時間、中央値は3時間である。10時間という値が平均を高くしている。0時間は有効な回答として含めた。この架空の5人の結果を、大学生全体の傾向と考えることはできない。
