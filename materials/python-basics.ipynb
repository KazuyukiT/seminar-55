### uvで授業用の環境を作る

**Windowsでは、uvをインストールしたWSLのLinux環境を使います。** VS Code本体はWindows側に置き、Microsoftの **WSL拡張**（`ms-vscode-remote.remote-wsl`）をインストールしてください。Macの人は通常のターミナルを使い、WSLの操作を省略します。

### 計算を表示する

````python
print(1200 + 800)
print(1200 * 3)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 3)
````

## リスト：複数の値をまとめる

5人の1週間の学習時間を、リストにまとめます。角括弧 `[]` で囲み、値をコンマで区切ります。この章のデータはすべて架空です。

````python
study_hours = [2, 5, 0, 3, 10]
print(study_hours[0])
print(study_hours[-1])
print(study_hours[1:3])
print(len(study_hours))
````

**結果：** `2`、`10`、`[5, 0]`、`5`。位置を表す番号（インデックス）は0から始まります。`-1` は最後、`[1:3]` は1番目から3番目の直前までです。`len()` は要素数を返します。

````python
total_hours = sum(study_hours)
mean_hours = total_hours / len(study_hours)
print(total_hours)
print(mean_hours)
print(min(study_hours), max(study_hours))
````

**結果：** `20`、`4.0`、`0 10`。`sum()` は合計、`min()` と `max()` は最小値・最大値です。平均は空のリストでは計算できません。

## 条件分岐：条件で処理を変える

````python
hours = 5
if hours >= 5:
    print("5時間以上")
else:
    print("5時間未満")
````

**結果：** `5時間以上`。`if` の後ろに条件を書き、行末に `:` を付けます。条件が成り立つと、その下の字下げした処理が実行されます。`else` はそれ以外です。字下げは半角スペース4個にそろえます。

| 比較 | 意味 |
|---|---|
| `a == b` | 等しい |
| `a != b` | 等しくない |
| `a > b`、`a >= b` | より大きい、以上 |
| `a < b`、`a <= b` | より小さい、以下 |

代入の `=` と比較の `==` を区別してください。条件を組み合わせるときは `and`（両方）、`or`（少なくとも一方）を使います。

````python
hours = 0
if hours == 0:
    print("0時間")
elif hours < 5:
    print("0時間より多く5時間未満")
else:
    print("5時間以上")
````

**結果：** `0時間`。`elif` は追加の条件です。上から調べ、最初に成り立った枝だけを実行します。この例は0以上の時間を前提にしています。

## 繰り返し：同じ処理を各データに行う

````python
study_hours = [2, 5, 0, 3, 10]
for hours in study_hours:
    print(hours * 60)
````

**結果（上から）：** `120`、`300`、`0`、`180`、`600`。`for` はリストから値を順に取り出して `hours` に代入し、字下げした処理を繰り返します。

### 条件に合う人数を数える

````python
count = 0
for hours in study_hours:
    if hours >= 5:
        count = count + 1
print(count)
````

**結果：** `2`。`count = count + 1` は、現在の値に1を足して代入し直します。`count` の初期化はループの前、最後の表示はループの外です。実行前に、2、5、0、3、10を読んだ直後の `count` を紙に書いてみてください。

## 辞書：項目名と値を対応させる

辞書は、項目名（キー）で値を取り出す仕組みです。リストの「何番目」と区別しましょう。

````python
student = {"id": "A", "hours": 2}
print(student["id"])
print(student["hours"])
````

**結果：** `A`、`2`。辞書をリストに入れると、小さな表のように複数人の情報を扱えます。

````python
students = [
    {"id": "A", "hours": 2},
    {"id": "B", "hours": 5},
    {"id": "C", "hours": 0},
]
for student in students:
    print(f"{student['id']}: {student['hours']}時間")
````

**結果：** `A: 2時間`、`B: 5時間`、`C: 0時間`。外側の二重引用符と内側の一重引用符を区別してください。大量の表データを扱うpandasは、この基礎の後で学びます。

## 関数：処理に名前を付ける

同じ計算を別の値でも使えるよう、関数にまとめます。

````python
def to_minutes(hours):
    return hours * 60

result = to_minutes(2.5)
print(result)
````

**結果：** `150.0`。`def` は関数の定義、`hours` は受け取る値の名前（引数）、`return` は返す値（戻り値）の指定です。定義しただけでは計算は実行されません。`to_minutes(2.5)` と呼び出すと動きます。

`print()` は表示、`return` は呼び出し元へ値を渡す操作です。返された値は次の計算に使えます。

````python
print(to_minutes(1) + to_minutes(2))
````

**結果：** `180`。関数内で `return` の代わりに `print` だけを使うと、呼び出しの戻り値は `None`（値がないことを表す値）になります。
